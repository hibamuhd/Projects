# Exploratory Analysis — Quick-Commerce Reorder & Cohort Dashboard

This notebook is a scratchpad version of `python/05_eda.py`, for interactively exploring the processed tables before finalizing the Power BI dashboard. Run the pipeline scripts (01-04) first so `data/processed/*.csv` exist.

```python
import pandas as pd
import matplotlib.pyplot as plt

fact_orders = pd.read_csv('../data/processed/fact_orders.csv')
dim_customer = pd.read_csv('../data/processed/dim_customer.csv')
dim_product = pd.read_csv('../data/processed/dim_product.csv')
cohort_retention = pd.read_csv('../data/processed/cohort_retention.csv')

fact_orders.head()
```

```python
# Overall reorder rate
print(f"Overall reorder rate: {fact_orders['reordered'].mean():.2%}")

# Customer segment breakdown
dim_customer['customer_segment'].value_counts(normalize=True)
```

```python
# Cohort retention curve
plt.figure(figsize=(8,5))
plt.plot(cohort_retention['order_number'], cohort_retention['retention_pct'], marker='o')
plt.xlabel('Order number (tenure)')
plt.ylabel('Retention %')
plt.title('Tenure-based cohort retention curve')
plt.grid(alpha=0.3)
plt.show()
```

```python
# Top 10 products by reorder rate (min 100 orders to avoid noise)
dim_product[dim_product['total_times_ordered'] >= 100].sort_values(
    'product_reorder_rate', ascending=False
).head(10)
```
